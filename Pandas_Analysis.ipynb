```python
import pandas as pd
import numpy as np

# Provided data
data = {
    'transaction_id': range(1, 21),
    'date': pd.date_range('2024-10-01', periods=20, freq='D'),
    'region': ['North', 'South', 'East', 'West', None, 'North', 'South', None, 'East', 'West',
               'North', 'South', 'East', 'West', 'North', None, 'East', 'West', 'North', 'South'],
    'product_category': ['Electronics', 'Clothing', None, 'Books', 'Electronics', 'Home',
                         'Clothing', 'Books', 'Electronics', None, 'Home', 'Clothing',
                         'Books', 'Electronics', 'Home', 'Clothing', 'Books', 'Electronics',
                         'Home', 'Clothing'],
    'sales_amount': [1200, 450, 890, None, 1500, 670, None, 340, 2100, 780,
                     560, None, 420, 1800, 920, 510, 380, None, 1100, 640],
    'quantity': [2, 5, 3, 1, None, 4, 2, 3, 1, 5,
                 3, 2, None, 1, 4, 3, 2, 1, None, 4],
    'customer_age': [25, 34, None, 45, 29, None, 38, 52, 27, 41,
                     33, None, 48, 26, 35, 42, None, 31, 39, 44],
    'payment_method': ['Credit Card', 'UPI', 'Cash', 'Debit Card', 'Credit Card',
                       'UPI', 'Cash', None, 'Credit Card', 'UPI', 'Debit Card',
                       'Cash', 'Credit Card', None, 'UPI', 'Cash', 'Debit Card',
                       'Credit Card', 'UPI', None]
}

df = pd.DataFrame(data)

# Display basic information
print("DataFrame Info:")
df.info()

print("\nMissing values per column:")
print(df.isna().sum())
```

Task 2: Handle Missing Values


```python
# 1. Region & Product_category: Fill with mode (most frequent value)
region_mode = df['region'].mode()[0]
df['region'].fillna(region_mode, inplace=True)

prod_mode = df['product_category'].mode()[0]
df['product_category'].fillna(prod_mode, inplace=True)

# 2. Sales_amount: Fill with median
sales_median = df['sales_amount'].median()
df['sales_amount'].fillna(sales_median, inplace=True)

# 3. Quantity: Fill using forward fill (ffill)
df['quantity'].fillna(method='ffill', inplace=True)

# 4. Customer_age: Fill with mean (rounded to integer)
age_mean = round(df['customer_age'].mean())
df['customer_age'].fillna(age_mean, inplace=True)

# 5. Payment_method: Drop rows where payment_method is missing
df.dropna(subset=['payment_method'], inplace=True)

# Verify no missing values remain
print("Missing values after cleaning:")
print(df.isna().sum())
```

Task 3: GroupBy Analysis


```python
# 1. Total sales by region
total_sales_region = df.groupby('region')['sales_amount'].sum()
print("Total sales by region:\n", total_sales_region)

# 2. Average sales by product_category
avg_sales_category = df.groupby('product_category')['sales_amount'].mean()
print("\nAverage sales by product category:\n", avg_sales_category)

# 3. Group by both region and product_category
grouped = df.groupby(['region', 'product_category']).agg(
    total_sales=('sales_amount', 'sum'),
    total_quantity=('quantity', 'sum')
).reset_index()
print("\nGrouped by region and category:\n", grouped)

# 4. Top 3 region-product combinations by sales
top3 = grouped.nlargest(3, 'total_sales')
print("\nTop 3 region-product combinations by sales:\n", top3)
```

Task 4: Custom Aggregation


```python
# Define function sales_range
def sales_range(x):
    return x.max() - x.min()

# Apply to find sales range for each region
region_sales_range = df.groupby('region')['sales_amount'].agg(sales_range)
print("Sales range by region:\n", region_sales_range)

# Use .agg() to calculate multiple metrics by region
region_metrics = df.groupby('region').agg(
    total_sales=('sales_amount', 'sum'),
    avg_sales=('sales_amount', 'mean'),
    max_sales=('sales_amount', 'max'),
    total_quantity=('quantity', 'sum'),
    min_quantity=('quantity', 'min')
)
print("\nMultiple metrics by region:\n", region_metrics)
```
